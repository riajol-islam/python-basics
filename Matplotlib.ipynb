```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Simple Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
import matplotlib.pyplot as plt
labels=["python","java","c++","javasript"]
sizes=[40,30,20,10]
plt.pie(sizes,labels=labels,autopct='%1.1f%%')
plt.title("Course distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 2, 5]
plt.bar(categories, values)
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['A', 'B', 'C', 'D']
values = [3, 7, 2, 5]
colors = ['red', 'green', 'orange', 'purple']

plt.bar(categories, values, color=colors)
plt.show()
```
