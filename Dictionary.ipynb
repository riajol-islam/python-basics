```python
data = { "name": "Jake", "age": 22 }
print(data)
```

```python
a = {"x": 1, "y": 2}
print(a)

b = dict(name="Sam", age=20)
print(b)
```

```python
d = {"name": "Kat", "age": 21}

print(d["name"])
print(d["age"])
```

```python
d = {"name": "Sam"}

d["age"] = 21        # Adding a new key-value pair
d["name"] = "Alex"   # Updating an existing value
print(d)
```

```python
d = {"a": 1, "b": 2}
del d["a"]
print(d)
```

```python
d = {"a": 1, "b": 2}
for key, value in d.items():
    print(key, value)
```
