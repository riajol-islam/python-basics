### creating array from scatch    

```python
zeros=np.zeros((4, 4))
print(zeros)
ones=np.ones((4, 4))
print(ones)

full=np.full((4, 4), 7)
print(full)

random=np.random.rand(4, 4)
print(random)

sequence=np.arange(0, 10, 2)
print(sequence)
```

### vector,amtrix and tensor

```python
vector=np.linspace(0, 1, 2)
print(vector)

matrix=np.eye(4)
print(matrix)

matrix_2=np.array([[1, 2], [3, 4]])
print(matrix_2)

tensor=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(tensor)
```

### array properties

```python
arr=np.array([1, 2, 3, 4, 5])
print(arr[0])
print("dimension:", arr.ndim)
print("shape:", arr.shape)
print("size:", arr.size)
print("data type:", arr.dtype)
print("item size:", arr.itemsize)
```

### array reshaping

```python
arr=np.arange(12)
print(arr)

reshaped_arr=arr.reshape(3, 4)
print(reshaped_arr)

reveled_arr=reshaped_arr.ravel()
print(reveled_arr)

transposed_arr=reshaped_arr.T
print(transposed_arr)
```
