```python
with open("data.txt","w") as file:
  file.write("hello world\n")
  file.write("hellow python!\n")
with open("data.txt","r") as file:
  print(file.read())
```

```python
with open("data.txt","a") as file:
  file.write("python needs to learn\n")
with open ("data.txt","r") as file:
  print(file.read())
```

```python
li=["mithi is now 3rd year mbbs student\n","she wants to neurosargeon\n","she loves to watch animie"]
with open ("data.txt","a") as file:
  file.writelines(li)
with open ("data.txt","r") as file :
  print(file.read())
```
