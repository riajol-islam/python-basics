# Advanved operations with Business Examples

```python
import numpy as np
import matplotlib.pyplot as plt

sales_data = np.array([
    [150000, 180000, 220000, 250000],
    [120000, 140000, 150000, 160000],
    [200000, 230000, 260000, 300000]
])
print("sales datashape:", sales_data.shape)
print("Sales data for Paradise Biryani:", sales_data[0])

plt.plot(sales_data[0], label='Paradise Biryani')
plt.legend()
plt.show()
```

```python
#total sales for each restaurant
total_sales = np.sum(sales_data, axis=1)
print("Total sales for each restaurant:", total_sales)

#total sales for par year
total_sales_per_year = np.sum(sales_data, axis=0)
print("Total sales for each year:", total_sales_per_year)
```

## Minimum sales per restaurant

```python
min_sales_per_year = np.min(sales_data, axis=0)
print("Minimum sales for each year:", min_sales_per_year)

##maximum sales for each year
max_sales_per_year = np.max(sales_data, axis=0)
print("Maximum sales for each year:", max_sales_per_year)


##Average for per restaurant
average_sales_per_restaurant = np.mean(sales_data, axis=1)
print("Average sales for each restaurant:", average_sales_per_restaurant)


```

## cumulative

```python
cumsum=np.cumsum(sales_data, axis=1)
print("Cumulative sales for each restaurant:", cumsum)

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Sales Over Time")
plt.xlabel("Quarters")
plt.ylabel("Cumulative Sales Amount")
plt.grid(True)
plt.show()
```
